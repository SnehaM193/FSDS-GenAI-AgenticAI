# Matplotlib Basics

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5,6,7,8]
y = [10,20,30,40,50,60,70,80]

plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,y, "bo--")
plt.show()
```

```python
x = [1,2,3,4,5]
y1 = [10,20,30,40,50]
y2 = [5,15,25,35,45]

plt.plot(x, y1)
plt.plot(x, y2)
plt.show()
```

```python
sales = [10,20,30,40,50]
profit = [5,15,20,30,45]

plt.plot(x, sales, label="Sales")
plt.plot(x, profit, label="Profit")

plt.legend(loc="upper left")
plt.show()
```

```python
x = [1,2,3,4,5,6,7,8]
y = [10,20,30,40,50,60,70,80]

plt.plot(x, y)
plt.grid()
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]

plt.plot(x, y)

plt.axis([0, 6, 0, 60])
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]

plt.plot(x, y, linestyle="-")
plt.plot(x, np.array(y) + 5, linestyle="--")
plt.plot(x, np.array(y) + 10, linestyle="-.")
plt.plot(x, np.array(y) + 15, linestyle=":")
```

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]
plt.plot(
    x,
    y,
    marker="o",
    linestyle="-",
    color="red",
    ms=8
)
plt.show()
```

```python
plt.scatter(x,y)
plt.xlabel("X")
plt.ylabel("Y")

plt.title("Scatter Plot")
plt.show()        
```

```python
marks = [40,34,89,54,62,65,57,90,70,55,85,34,75]

plt.hist(marks, bins=5)
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Marks Distribution")
plt.show()
```

```python
subject = ["Java","Python","SQL","C++"]
marks = [67,89,95,75]

plt.bar(subject,marks)
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.title("Subject Marks")
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]

error = [1,2,1,3,2]
plt.errorbar(x, y, yerr = error, fmt='o-',ecolor='red')
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Error Bar Chart")
plt.show()
```

```python
values = [35, 25, 20, 45]
labels = ["Java","Python","SQL","C++"]

plt.pie(values, labels=labels, autopct="%1.1f%%")
plt.title("Programming Languages")
plt.show()
```

```python
x_area = np.arange(1,6)
y_area = np.array([1,4,6,8,4])

plt.fill_between(x_area, y_area)
plt.plot(x_area, y_area)
plt.title("Area Chart")
plt.show()
```

```python
x = np.linspace(0,10,50)

plt.plot(x, np.sin(x), label="sin(x)")
plt.plot(x, np.cos(x), label="cos(x)")

plt.legend()
plt.grid()
plt.show()
```

# Practical Example

```python
seasons = np.array([2019,2020,2021,2022,2023,2024])

player1 = np.array([450, 500, 550, 480, 600, 650])
player2 = np.array([400, 450, 500, 520, 580, 620])

plt.figure(figsize=(8,5))

plt.plot(
    seasons,
    player1,
    linestyle="-.",
    marker="d",
    ms=7,
    label="Player1"
)
plt.plot(
    seasons,
    player2,
    linestyle="--",
    marker="o",
    ms=7,
    label="Player2"
)

plt.xlabel("Seasons")
plt.ylabel("runs")
plt.title("Players Performance Across Seasons")

plt.legend()
plt.grid()
plt.show()
```
