# Data Cleaning

```python
emp['Name']
```

```python
emp['Name'] = emp['Name'].str.replace(r'\W','',regex=True)           # Remove special characters, spaces, and symbols from the Salary column
```

```python
emp['Name']
```

```python
emp['Domain']
```

```python
emp['Domain'] = emp['Domain'].str.replace(r'\W','', regex=True)
```

```python
emp['Domain']
```

```python
emp['Age']
```

```python
emp['Age'] = emp['Age'].str.extract('(\d+)')
```

```python
emp['Age']
```

```python
emp['Location'] = emp['Location'].str.replace(r'\W','',regex=True)
```

```python
emp['Location']
```

```python
emp['Salary'] = emp['Salary'].str.replace(r'\W','',regex=True)
```

```python
emp['Salary']
```

```python
emp['Exp'] = emp['Exp'].str.extract('(\d+)')         # Extract one or more digits from the Exp column, removing text like 'years'
```

```python
emp
```

```python
clean_data = emp.copy()
```
