# Data visualization

```python
import seaborn as sns
```

```python
sample_ratings = ratings.sample(n=100000, random_state=42)

sns.histplot(data=sample_ratings, x='rating', bins=10)
plt.title('Distribution of Movie Ratings')
plt.xlabel('Ratings')
plt.ylabel('Frequency')
plt.show()
```

```python
sns.boxplot(data=sample_ratings, x='rating')
plt.show()
```

```python
rating_count = ratings.groupby('rating')[['movieId']].count()
rating_count
```

```python
avg_ratings = ratings.groupby("movieId", as_index=False)["rating"].mean()
avg_ratings.head()
```

```python
movie_ratings = movies.merge(avg_ratings, on="movieId", how="inner")
movie_ratings.tail()
```
