```python
arr1 = np.array([[1,2],[3,4]])
arr1
```

# Mathematical Functions

```python
#Square Root
arr = np.array([4,9,25,64])
square_root = np.sqrt(arr)
print(square_root)
```

```python
#Absolute Value
arr = np.array([-10,20,30,-40])
absolute_value = np.abs(arr)
print(absolute_value)
```

```python
arr = np.array([2.434, 4.567])
print(np.round(arr))
```

```python
arr = np.array([4,9,25,64])
print(np.power(arr, 2))
```

# Statistical And Aggregation Functions

```python
marks = np.array([70,50,30,20,40])
print(np.sum(marks))
print(np.mean(marks))
print(np.median(marks))
print(np.var(marks))
print(np.std(marks))
```

```python
print(np.min(marks))
print(np.max(marks))
```

```python
marks = np.array([[1,2,3],[4,5,6]])
print(np.sum(marks, axis=0))
print(np.sum(marks, axis=1))
```
