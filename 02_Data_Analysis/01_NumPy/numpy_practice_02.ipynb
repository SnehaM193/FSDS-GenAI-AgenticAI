# Boolean Indexing, Masking and Filtering

```python
import numpy as np
```

```python
arr = np.array([10,20,30,40,50])
print(arr > 30)
```

```python
marks = np.array([45,67,89,92])
result = marks[(marks > 50) & (marks <= 90)]
print(result)
```

```python
#To find positions based on condition
np.where(marks >= 50)
```

```python
marks = np.array([45,67,89,92])
result = np.where(marks >= 50, "Pass","Fail")
print(result)
```

# Array Manipulation
changing structrue or form of an array

```python
arr = np.array([1,2,3,4,5])
print(arr)
new_arr= arr.reshape(5,1)
print(new_arr)
```

```python
#flatten: converts multidimensional array into 1D array  
arr = np.array([[1,2,3],[4,5,6]])
print(np.ravel(arr))
```

```python
#Transpose: changes rows into columns and columns into rows
print(np.transpose(arr))
```

```python
#Concatenate: joining two arrays
arr = np.array([1,2])
arr1 = np.array([2,3])
np.concatenate((arr,arr1))
```

```python
#joining two arrays vertically
arr = np.array([1,2])
arr1 = np.array([2,3])
np.vstack((arr,arr1))
```

```python
#joining two arrays horizontally
arr = np.array([1,2])
arr1 = np.array([2,3])
np.hstack((arr,arr1))
```

```python
#split array
arr = np.array([1,2,3,4])
np.split(arr,2)
```

# Sorting And Searching

```python
arr = np.array([40,50,30,20,90])
np.sort(arr)
```

```python
np.argsort(arr)     #return the indexes that would sort the array
```

```python
arr = np.array([23,25,67,89,23])
np.unique(arr)
```

```python
np.searchsorted(arr,28)         #finds position where value can be inserted while keeping the array sorted
```

# SET Operations

```python
a = np.array([1,2,3,4])
b = np.array([3,4,7,8])
np.union1d(a, b)              #returns all unique values from both the arrays
```

```python
print(np.intersect1d(a, b))   #returns common values
```

```python
print(np.setdiff1d(a,b))       #returns values present in first array not in second array
```

# Handling Missing and Special Values

```python
#np.nan represents missingor undefined value
arr = np.array([10,20,30,np.nan,40])
np.isnan(arr)
```

```python
#np.inf represents infinity
arr = np.array([10,np.inf,40])
np.isinf(arr)
```
