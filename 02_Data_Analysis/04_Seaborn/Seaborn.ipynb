# Seaborn: Data Visualization
Seaborn is a Python library used for data visualization. It helps us create attractive and informative statistical graphs to understand data easily.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_theme(style="whitegrid")

%matplotlib inline
plt.rcParams["figure.figsize"] = (6,2)

import warnings
warnings.filterwarnings('ignore')
```

# Univariate Analysis
Univariate Analysis means analyzing one variable at a time; a Histogram shows the distribution of numerical data, while the KDE curve gives a smooth view of its shape.

```python
sns.histplot(data=df, x="InternetUsers",bins=15,kde=True)
plt.title("Distribution of Internet Users")
plt.xlabel("Internet Users")
plt.ylabel("Frequency")
plt.show()
```

# Count Plot
Count Plot in Seaborn is used to show the number of times each category appears in a categorical column

```python
sns.countplot(data=df, x="IncomeGroup", order=df["IncomeGroup"].value_counts().index)
plt.title("Number of Records By Income Group")
plt.xlabel("IncomeGroup")
plt.ylabel("Count")
plt.xticks(rotation=25, ha="right")
plt.show()
```

# Bivariate Analysis
Bivariate Analysis means analyzing two variables (two columns) at a time to understand the relationship or connection between them.

```python
sns.scatterplot(data=df, x="InternetUsers", y="BirthRate")
plt.title("Internet Users Vs Birth Rate")
plt.xlabel("Internet Users")
plt.ylabel("Birth Rate")
plt.show()
```

# Scatter Plot with Trend line
The regression line summarizes the overall linear trend.

```python
sns.regplot(data=df, x="InternetUsers", y="BirthRate",
          scatter_kws={"alpha":0.6}, line_kws={"color":"red"})
plt.title("Internet Users Vs Birth Rate with trend line")
plt.show()
```

# Multivariate Analysis
Multivariate Analysis means analyzing three or more variables (columns) at the same time to understand the relationships and patterns among them.

```python
sns.scatterplot(data=df, x="InternetUsers", y="BirthRate", hue="IncomeGroup")
plt.title("Internet Users Vs Birth Rate by Income group")
plt.xlabel("Internet Users")
plt.ylabel("Birth Rate")
plt.show()
```

# Box Plot and Outliers
A graph used to show the distribution of numerical data, including median, quartiles, and outliers.Points beyond whiskers may be the potential outliers.

```python
sns.boxplot(data=df, x="IncomeGroup", y="BirthRate")
plt.title("Birth Rate by Income Group")
plt.xlabel("Income Group")
plt.ylabel("Birth rate")
plt.xticks(rotation=25, ha="right")
plt.show()
```

# Violin Plot
A graph that shows the distribution and density of numerical data across different values

```python
sns.violinplot(data=df, x="IncomeGroup", y="BirthRate", inner="quartile", cut=0)
plt.title("Birth Rate distribution by Income Group")
plt.xlabel("Income Group")
plt.ylabel("Birth rate")
plt.xticks(rotation=25, ha="right")
plt.show()
```

# Bar Plot
A bar plot is a graph that uses rectangular bars to compare values between different categories.

```python
sns.barplot(data=df, x="IncomeGroup", y="BirthRate", errorbar=("ci",95))
plt.title("Mean Birth Rate by Income Group")
plt.xlabel("Income Group")
plt.ylabel("Mean Birth rate")
plt.xticks(rotation=25, ha="right")
plt.show()
```

# Correlation Heatmap
A correlation heatmap is a graph that shows the relationship between different numerical columns in a dataset using colours and values.

```python
numeric_df = df.select_dtypes(include="number")
correlation = numeric_df.corr()

sns.heatmap(correlation, annot=True, cmap="coolwarm", center=0, fmt=".2f")
plt.title("Correlation heatmap of numeric variables")
plt.show()
```

# Pair Plot
A Pair plot displaye pair wise relationships betweem selected numeric variables and their individual distribution.

```python
sns.pairplot(
    df,
    hue='IncomeGroup',
    palette='rainbow',
    diag_kind='kde',
    kind='scatter'
)

plt.show()
```
