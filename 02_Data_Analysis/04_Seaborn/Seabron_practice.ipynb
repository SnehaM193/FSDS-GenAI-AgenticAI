```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.DataFrame({
    'Student': ['Rahul', 'Priya', 'Amit', 'Sneha', 'Riya',
                'Karan', 'Anjali', 'Rohan', 'Neha', 'Arjun'],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Female',
               'Male', 'Female', 'Male', 'Female', 'Male'],
    'Study_Hours': [2, 5, 3, 7, 6, 4, 8, 3, 6, 5],
    'Marks': [45, 80, 55, 90, 85, 60, 95, 50, 88, 75],
    'Attendance': [60, 90, 70, 95, 85, 75, 98, 65, 92, 80],
    'Age': [20, 21, 20, 22, 21, 20, 22, 21, 20, 22]
})
df
```

```python
sns.histplot(data=df,x="Marks",kde=True)  #Univariate analysis
plt.show()
```

```python
sns.countplot(data=df,x="Gender")
plt.show()
```

```python
sns.scatterplot(data=df,x="Study_Hours",y="Marks")            #bivariate analysis
plt.show()
```

```python
sns.boxplot(data=df,x="Gender",y="Marks")
plt.show()
```

```python
sns.scatterplot(data=df,x="Study_Hours",y="Marks",hue="Gender")
plt.show()
```

```python
sns.pairplot(df,
    hue="Gender",
    diag_kind='kde',
    kind='scatter'
)
plt.show()
```

```python
numeric_df = df.select_dtypes(include="number")
correlation = numeric_df.corr()

sns.heatmap(correlation, annot=True, cmap="coolwarm", center=0, fmt=".2f")
plt.show()
```
