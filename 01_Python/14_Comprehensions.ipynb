# Comprehensions

# List

```python
numbers = [2,3,4,5,6,8,9]
squares = [x ** 2 for x in numbers]
print(squares)
```

```python
numbers = [12,22,45,78,34]
even_numbers = [x for x in numbers if x % 2 == 0]
print(even_numbers)
```

```python
numbers = [22,45,68,34,12]
result = ['Even' if x % 2 == 0 else 'odd' for x in numbers]
print(result)
```

```python
words = ['python','java','sql']
upper_case = [word.upper() for word in words]
print(upper_case)
```

```python
numbers = [1,2,3]
letters = ["A","B"]
result = [(number,letter) for number in numbers for letter in letters ]
print(result)
```

# Dictionary

```python
numbers = [1,2,3,4,5]
squares = {(x**2) for x in numbers}
print(squares)
```

```python
numbers = [1, 2, 3, 4, 5, 6]
even_numbers = {x: x ** 2 for x in numbers if x % 2 == 0}
print(even_numbers)
```

```python
numbers = [12,34,22,67,75,98]
result = {x: 'Even' if x % 2 == 0 else 'odd' for x in numbers}
print(result)
```

```python
prices = {'apple' : 100,'banana':200,'orange':400}
new_prices = {key: value * 2 for key,value in prices.items()}
print(new_prices)
```
