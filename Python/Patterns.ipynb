```python
#Printing patterns in python
```

```python
for i in range(4):
    for j in range(4):
        print('*', end = ' ')
    print()
```

```python
for i in range(4):
    for j in range(i+1):
        print('*', end=' ')
    print()
```

```python
for i in range(4):
    for j in range(1,5):
        if i < j:
            print('*', end = ' ')
    print()
```

```python
for i in range(5):
    for j in range(2*i+1):
        print('*', end=' ')
    print()
```

```python
for i in range(5,0,-1):
    print(''*(5-i)+' * ' * (2*i-1))
```

```python
for i in range(5):
    for j in range(5):
        if i == 0 or i == 4 or j == 0 or j == 4:
            print('*', end=' ')
        else:
            print(' ', end= ' ')
    print()
```

```python
for i in range(5):
    for j in range(i+1):
        print(j+1, end=' ')
    print()
```

```python
num = 1
for i in range(1,6):
    for j in range(1,i+1):
        print(num, end = ' ')
        num+= 1
    print()
```

```python
for i in range(1,6):
    for j in range(1,i+1):
        if j == 1 or j == i or i == 5:
            print('*', end=' ')
        else:
            print('', end = '')
    print()
        
```

```python
for i in range(1,6):
    for j in range(i):
        print(chr(65+j), end= ' ')
    print()
```

```python
for i in range(1, 5):
    print("  " * (4 - i) + "* " * (2 * i - 1))
```
