```python
temp = int(input('Enter a temp: '))

if temp < 20:
    print('Temperature is cold')
elif temp >= 20 and temp <= 35:
    print('Temperature is moderate')
elif temp > 35:
    print('Temperature is hot')

else:
    print('Invalid')
```

```python
purchase_amt = int(input('Enter Purchase amount: '))

if purchase_amt < 3000:
    print('No discount')
elif purchase_amt > 3000 and purchase_amt <= 4999:
    print('10% discount')
elif purchase_amt >= 5000:
    print('20% discount')
else:
    print('amount is less for discount')
```
