```python
i = 1

while i<=5:
    print('data science') # when we mention end then new line will not create
    
    j = 1
    while j<=4:
        print('technology')
        j = j + 1
        
    i = i + 1
    print()
    
```

```python
i = 1

while i <= 5:
    print('Data Science')

    j = 1
    while j <= 4:
        print('technology')
        j = j + 1

    i = i + 1
    print()
```

```python
i = 1

while i <= 5:
    print('Data Science', end = '-')

    j = 1
    while j <= 4:
        print('technology', end = '-')
        j = j + 1

    i = i + 1
    print()
```
