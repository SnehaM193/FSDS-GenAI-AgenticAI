```python
first_name = 'Sneha'
last_name = 'Mohanty'
Country = 'India'
State = 'Maharashtra'
age = 22
skills = ['HTML','CSS','JS','PYTHON']
personal_info = {
  'firstname':'Sneha',
  'lastname':'Mohanty',
  'country': 'India'
}

#printing all the values stored in variables

print("first name:",first_name)
print("first name length:",len(first_name))
print("last name:",last_name)
print("last name length:", len(last_name))
print("Country name:",Country)
print("State name:", State)
print("age:", age)
print("skills:", skills)
print("Personal info",personal_info)
```

```python
#declaring multiple variables in single line
first_name, last_name, country, state = 'Sneha', 'Mohanty', 'India', 'Maharashtra'

print(first_name, last_name, country, state)
print('First Name:', first_name),('Last Name:', last_name),('Country:', country),

print('First Name:', first_name)
print('Last Name:', last_name)
print('Country:', country)
print('State:', state)
```
